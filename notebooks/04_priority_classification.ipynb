```python
df_priority = pd.read_csv("../results/patient_predictions.csv")

df_priority.head()
```

```python
def assign_priority(probability):
    if probability >= 0.70:
        return "High"
    elif probability >= 0.40:
        return "Medium"
    else:
        return "Low"
```

```python
df_priority["Followup_Priority"] = df_priority[
    "Readmission_Probability"
].apply(assign_priority)

df_priority.head()
```

```python
priority_counts = df_priority["Followup_Priority"].value_counts()

print(priority_counts)

print(
    df_priority["Followup_Priority"]
    .value_counts(normalize=True) * 100
)
```

```python
df_priority.to_csv(
    "../results/patient_priority_results.csv",
    index=False
)

print("Priority classification results saved successfully!")
```

```python
df_priority.insert(
    0,
    "Patient_ID",
    ["P" + str(i).zfill(5) for i in range(1, len(df_priority) + 1)]
)

df_priority.head()
```
